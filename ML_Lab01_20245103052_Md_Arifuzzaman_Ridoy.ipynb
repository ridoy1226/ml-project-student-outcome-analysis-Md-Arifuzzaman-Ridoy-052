```python

import numpy as np

np.random.seed(42)

marks = np.array([55, 72, 48, 90, 66, 81])

print("Marks:", marks)
print("Mean:", marks.mean())
print("Highest:", marks.max())
print("Passed:", marks[marks >= 50])

# 12. Find minimum mark and standard deviation
print("Minimum:", marks.min())
print("Standard Deviation:", marks.std())

# 13. Add 5 bonus marks without changing original array
bonus_marks = marks + 5
print("Marks with Bonus:", bonus_marks)
print("Original Marks:", marks)

# 14. Count how many students passed
passed_count = np.sum(marks >= 50)
print("Number of Passed Students:", passed_count)

# 15. Create a 2 × 3 matrix from the marks
matrix = marks.reshape(2, 3)
print("2 × 3 Matrix:")
print(matrix)
```

```python
import pandas as pd

# Create the dataset
data = {
    'student': ['Ahsan', 'Babon', 'Chayon', 'Dipu', 'Rana', 'Joy'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

# Original DataFrame
print("Original DataFrame:")
print(df)

# Basic Information
print("\nDataFrame Info:")
df.info()

# Descriptive Statistics
print("\nDescriptive Statistics:")
print(df.describe())

# 16. Which columns are features and which column is the label?
features = ['study_hours', 'sleep_hours', 'attendance', 'snacks']
label = 'passed'

print("\n16. Features:", features)
print("Label:", label)

# 17. How many samples and columns are present?
rows, columns = df.shape

print("\n17. Number of Samples:", rows)
print("Number of Columns:", columns)

# 18. Filter students with attendance of at least 80
attendance_80 = df[df['attendance'] >= 80]

print("\n18. Students with attendance >= 80:")
print(attendance_80)

# 19. Calculate average study hours for students who passed
passed_students = df[df['passed'] == 'Yes']
average_study_hours = passed_students['study_hours'].mean()

print("\n19. Average study hours of passed students:", average_study_hours)

# 20. Sort DataFrame by attendance in descending order
sorted_df = df.sort_values(by='attendance', ascending=False)

print("\n20. DataFrame sorted by attendance (descending):")
print(sorted_df)

# 21. Does this dataset prove that snacks cause failure?
print("\n21. Conclusion:")
print("No, this tiny dataset does not prove that snacks cause failure.")
print("It only shows an association in these few observations.")
print("To prove causation, we would need more data and a controlled study.")
```

```python

import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)

    return np.random.randint(40, 101, size=5)


print("Seeded run 1:", generate_scores(True))
print("Seeded run 2:", generate_scores(True))
print("Unseeded run 1:", generate_scores(False))
print("Unseeded run 2:", generate_scores(False))
```
